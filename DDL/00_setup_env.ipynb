#  **_We proceed to create a catalog_**
File 00_setup_env will be the starting point from where we config our environment to have bronze layer with Catalogs, schemas and tables.

```python
%sql
/* Create Catalog */

CREATE CATALOG IF NOT EXISTS bootcamp
```

```python
%sql

/* Using the catalog */
USE CATALOG bootcamp

/* Schema creation */
CREATE SCHEMA IF NOT EXISTS bootcamp.raw

/* We can check all this with SHOW SCHEMAS */
SHOW SCHEMAS
```

## **_Volume creation and dataset location_**

```python
%sql
/* Once we create the volume, it is a good practice to comment it */
CREATE VOLUME IF NOT EXISTS bootcamp.raw.archivos COMMENT 'Raw CSV file';

/* Check it if this was created */
SHOW VOLUMES IN bootcamp.raw
```

**At this point we need for this project to upload the CSV file. 
We will put it in the "archivos" volume from the schema "raw" we created**

```python
%sql
/* And we proceed to create the bronze layer table */
CREATE TABLE IF NOT EXISTS bootcamp.raw.propiedades_bronze

/* We can use the read_files function to read the csv file */
SELECT * FROM read_files(
  '/Volumes/bootcamp/raw/archivos/properties_bronze.csv',
  format => 'csv',
  header => true
  )
WHERE url LIKE 'https%'

/* 
NOTE: Why do we need url like? --> Since this is a real state project the most important field to validate is the url, why? Because we need to know if the url is valid or not, if it is not valid we can discard the record, if it is valid we can proceed to the next step. 
*/
```

## Take in mind the below since you can review the scripts you run

```python
%sql 

/* Check the table */
DESCRIBE EXTENDED bootcamp.raw.propiedades_bronze

/* Or in any case you can check if the table was created with :*/
SHOW TABLES IN bootcamp.raw
```
